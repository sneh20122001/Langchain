## What is Langchain??

LangChain is an open-source software framework that makes it easier to build applications using large language models (LLMs) like GPT-4 or Llama.It's provide ready to use libraries for custom use of chunking, model integration, vectorization, embedding, llm, and many more. Overall, LangChain provides plug-and-play wrappers for all of it.


## What is an Agent??

An AI agent is an LLM given tools (like a search engine, code interpreter, or SQL database) and the authority to make decisions in a loop. Instead of answering once, it looks at a task, decides which tool to use, checks the tool's output, and repeats until the goal is finished.

## What is LangGraph??

LangGraph provides the orchestration engine for agents. It maps out agent workflows as nodes  and edges. Crucially, it manages cycles and persistent state, allowing agents to pause for human input or recover after a system crash.

## What is LangSmith??

LangSmith is the observability layer. It logs every API call, latency metric, token count, and step your agent takes. If an agent goes off track or generates bad code, LangSmith shows you exactly which prompt or tool call caused the issue.

## What is Deep Agents??

A deep agent is an advanced AI agent architecture designed to handle complex, long-horizon, multi-step tasks autonomously by incorporating structured planning, persistent memory, and subagent delegation.Example : Coding agent (review code, identify tools, wait for human input,and many more.)

                            │
       │   LangGraph (State & Workflow Engine)    │  <- "Road Network & Traffic Rules"

       │   Runtime / Persistence / Execution      │  <- "Power, Storage & Highway Infrastructure"
                            │
       │   LangSmith (Observability & Monitoring) │  <- "Control Room & Dashcams"

## Why are LangChain agents built on top of LangGraph?

1. Simple chains failed in production: Early agents operated in linear pipelines (Step A → Step B → Step C). If Step B failed, the whole process crashed.
2. Real tasks require loops & memory: Real-world workflows require an agent to loop back, retry failed steps, ask humans for approval, and retain memory across restarts.  
3. Separation of concerns: LangChain acts as the components (the tools and LLM wrappers). LangGraph acts as the engine that runs those components reliably in a stateful loop
