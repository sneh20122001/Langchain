# LangChain Providers and Models



## 1. The Big Picture

**LangChain** gives you one common way to work with chat models from different AI providers.

Think of it like a **universal adapter**:

- Your application talks to LangChain.
- LangChain talks to the selected provider.
- The provider runs the AI model.
- The model sends a response back through the same path.

This means your application code can often stay similar even when you change providers.


## 2. What is a Provider?

A **provider** is a company or platform that hosts AI models and gives you access to them through an API.

Examples from the source material include:

- OpenAI
- Anthropic
- Google
- AWS Bedrock

A useful distinction:

**Provider = who gives you access to the model**

**Model = the particular AI model you ask that provider to run**

For example, `openai:gpt-5.5` contains both pieces of information: `openai` identifies the provider and `gpt-5.5` identifies the model.


## 3. Why Does LangChain Use Provider Packages?

Each provider can have different APIs, authentication methods, and special features. LangChain uses **dedicated integration packages** to connect its common interface to those providers.

Examples:

- `langchain-openai` → OpenAI integration
- `langchain-anthropic` → Anthropic integration
- `langchain-google-genai` → Google integration

The important idea is **separation**: your application can use LangChain's standard model interface while the provider-specific package handles communication with that provider.


## 4. One Common Interface

LangChain's chat models implement a common interface. This gives your code familiar operations such as:

- `invoke()` — send a request and get a response
- `stream()` — receive a response progressively
- `bind_tools()` — connect tools to a model when supported

The benefit is that you can **swap providers without rewriting your whole application**.

You can also compare models using similar application code and use advanced capabilities such as tool calling, structured output, and streaming where supported.


## 5. The Easiest Way to Initialize a Model

LangChain provides `init_chat_model()` as a convenient way to initialize a chat model.

The format is:

```text
provider:model
```

So:

```text
openai:gpt-5.5
```

means: use the OpenAI integration and the `gpt-5.5` model.


````python
from langchain.chat_models import init_chat_model

openai_model = init_chat_model("openai:gpt-5.5")
anthropic_model = init_chat_model("anthropic:claude-opus-4-8")
google_model = init_chat_model("google-genai:gemini-3.1-pro-preview")
````

### What did we create?

We created three model objects. Each object represents a different provider/model combination.

The application can then use the same general operation on each model.


````python
for model in [openai_model, anthropic_model, google_model]:
    response = model.invoke("Explain quantum computing in one sentence.")
    print(response.text)
````

## 6. What Does `invoke()` Mean?

In simple terms, `invoke()` means **send this input to the model and wait for the model's result**.

For example:

```python
response = model.invoke("Hello!")
```

Here:

1. Your Python code sends the message.
2. LangChain uses the selected integration.
3. The provider receives the request.
4. The model generates a response.
5. The response is returned to your Python program.

The exact response structure and supported capabilities can depend on the model/integration.


## 7. Two Ways to Specify a Model

### Method A — Provider prefix

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("openai:gpt-5.5")
```

This is useful because the provider is explicit.

### Method B — Direct class

```python
from langchain_openai import ChatOpenAI
model = ChatOpenAI(model="gpt-5.5")
```

The direct class approach is useful when you want to work directly with a provider-specific integration.


## 8. API Keys and Environment Variables

AI providers generally require credentials so their API knows **who is making the request** and can apply the appropriate account, permissions, and billing.

The source material notes that LangChain integrations support automatic API-key handling through environment variables.



## 9. Finding Model Names

Different providers support different model names. You should check the provider's documentation for the models currently available to you.

The source lists model documentation for OpenAI, Anthropic, Google AI, AWS Bedrock, Ollama, and Groq.

A key lesson: **the provider and model name must match what the integration and provider API actually support.**


## 10. Routers and Proxies

A **router/proxy** can sit between your application and multiple model providers.

Instead of your application separately managing every provider, a router can give you one interface or credential and route requests to different models.

The source mentions examples such as OpenRouter and LiteLLM.

Routers can be useful for:

- accessing models from many providers
- switching models more easily
- fallback behavior when a primary model fails
- routing and load balancing
- centralized spend tracking, depending on the router

**Mental model:**

`Your application → Router → Selected provider/model`


## 11. OpenAI-Compatible Endpoints

Some providers expose APIs that are compatible with the OpenAI API style.

LangChain's `ChatOpenAI` can be configured with a custom `base_url` for such endpoints.

Conceptually:

```text
ChatOpenAI
   ↓
custom base URL
   ↓
OpenAI-compatible provider
   ↓
model
```

However, compatibility does not necessarily mean every provider-specific feature is identical. The source specifically warns that `ChatOpenAI` targets official OpenAI API specifications and may not preserve non-standard response fields from third-party providers.


## 12. Deep Understanding: How the Pieces Fit Together

Imagine you are building a chatbot.

### Without a common abstraction

Your application might need separate code for every provider:

`Application → OpenAI-specific code`

`Application → Anthropic-specific code`

`Application → Google-specific code`

Changing providers could require changes throughout your application.

### With LangChain's common interface

`Application → LangChain interface → Provider integration → Model`

Your application focuses more on **what it wants the model to do**, while the integration layer handles provider-specific communication.

That abstraction is the main reason the provider/model architecture is useful.


## 13. Important Terms to Remember

| Term | Simple meaning |
|---|---|
| Provider | Company/platform that hosts AI models and provides API access |
| Model | Specific AI model you call |
| Integration package | Code that connects LangChain to a provider |
| `init_chat_model()` | Convenient way to initialize a chat model |
| `invoke()` | Send an input and receive a model response |
| Router | Layer that can route requests among multiple providers/models |
| Proxy | An intermediary endpoint between your app and model service |
| `base_url` | The API endpoint address used by a client |
| Environment variable | A safer place to provide configuration/secrets outside source code |


## 14. Quick Revision

Remember this chain:

**Application → LangChain → Integration → Provider → Model → Response**

And remember these three ideas:

1. **Provider** tells you where the model service comes from.
2. **Model** tells you which AI model you want to use.
3. **Common interfaces** make it easier to change or compare providers.

For GitHub, keep your notebook readable and **never commit real API keys or other secrets**.
